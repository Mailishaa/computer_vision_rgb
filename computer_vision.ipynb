- **Output:** Predicted class or probability for each condition
- **Goal:** Support medical image analysis and research, not replace clinical diagnosis

## STEP 5

In this step, I separated the image into its three main color channels: **Red, Green, and Blue (RGB)**.

### 1. Extracting the color channels


```python

r = image[:,:,0]
g = image[:,:,1]
b = image[:,:,2]

f, (ax1,ax2,ax3,) = plt.subplots(1,3,figsize= (20,10))
ax1.set_title('R channel')
ax1.imshow(r,cmap='gray')
ax2.set_title('G channel')
ax2.imshow(g,cmap='gray')
ax3.set_title('B channel')
ax3.imshow(b,cmap='gray')
```

```python
hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)
h = hsv[:,:,0]
s = hsv[:,:,1]
v = hsv[:,:,2]

f, (ax1,ax2,ax3,) = plt.subplots(1, 3,figsize= (20,10))
ax1.set_title('Hue')
ax1.imshow(h, cmap='gray')

ax2.set_title('Saturation')
ax2.imshow(s, cmap='gray')

ax3.set_title('Value')
ax3.imshow(v, cmap='gray')
```
